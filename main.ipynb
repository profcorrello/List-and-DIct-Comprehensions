## 📝 Module: `list_comprehensions.py`

### Purpose

Demonstrates the two most common uses of list comprehensions — **filtering** and **mapping** — by showing the traditional `for` loop side by side with the comprehension version.

### Public API

#### `demonstrate_list_comprehension(data: List[str]) -> None`

| Argument | Type | Description |
| --- | --- | --- |
| `data` | `List[str]` | Strings to filter and transform. |

Returns `None`; all results are printed.

### Concepts shown

**1. Filtering** — select items that satisfy a condition:

```python
# Traditional loop
filtered: List[str] = []
for item in data:
    if len(item) > 5:
        filtered.append(item)

# Comprehension  →  [expression for item in iterable if condition]
filtered = [item for item in data if len(item) > 5]
```

**2. Mapping** — apply an expression to every item:

```python
# Traditional loop
upper: List[str] = []
for item in data:
    upper.append(item.upper())

# Comprehension  →  [expression for item in iterable]
upper = [item.upper() for item in data]
```

### Design notes

- Both versions are shown for each demo so you can compare line count and readability directly.
- Use a plain `for` loop when the body has multiple statements or side effects (logging, printing, I/O).

````python
from list_comprehensions import demonstrate_list_comprehension

demonstrate_list_comprehension(WORD_LIST)
````

## 🗂️ Module: `dict_comprehensions.py`

### Purpose

Shows how to build a dictionary from an iterable of key-value pairs in a single line, replacing the initialize-then-assign loop pattern.

### Public API

#### `demonstrate_dict_comprehension(data: List[Tuple[str, int]]) -> None`

| Argument | Type | Description |
| --- | --- | --- |
| `data` | `List[Tuple[str, int]]` | `(key, value)` tuples to place into the dictionary. |

Returns `None`; results are printed.

### Concepts shown

```python
# Traditional loop: initialize, then assign each pair
student_scores: Dict[str, int] = {}
for student, score in data:
    student_scores[student] = score

# Dictionary comprehension: one expression, one line
# Syntax: {key_expression: value_expression for item in iterable}
student_scores = {student: score for student, score in data}
```

Because each element of `data` is a 2-tuple, it unpacks directly into `student, score` in the `for` clause.

### Design notes

- Conditions work the same way as in list comprehensions: `{student: score for student, score in data if score >= 90}`
- Duplicate keys keep the last value, just like with loop assignment.

````python
from dict_comprehensions import demonstrate_dict_comprehension

demonstrate_dict_comprehension(SCORE_DATA)
````
